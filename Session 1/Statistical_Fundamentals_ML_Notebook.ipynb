## Part 2: Population vs Sample
A population is the complete set of observations. A sample is a subset used for analysis.

## Part 3: Mean, Median and Mode

```python
data = np.array([4,5,1,2,7,2,6,9,3])
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Mode:", stats.mode(data).mode)
```

## Part 5: Variance and Standard Deviation

```python
from statistics import variance, stdev
game_points=[35,56,43,59,63,79,35,41,64,43,93,60,77,24,82]
print(variance(game_points))
print(stdev(game_points))
```

## Part 8: Correlation

```python
hours_studied=[1,2,3,4,5,6,7]
exam_scores=[45,50,58,62,70,81,90]
print(np.corrcoef(hours_studied,exam_scores))
```
