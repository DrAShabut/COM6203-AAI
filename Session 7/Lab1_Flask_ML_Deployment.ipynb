## Part 2: Import Libraries

```python
# Data manipulation libraries
import numpy as np
import pandas as pd

# Model persistence
import joblib

# Machine learning libraries
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

## Part 3: Load the Dataset
We will use the California Housing dataset included with Scikit-learn.


```python
# Load California Housing dataset
housing = fetch_california_housing(as_frame=True)

# Convert to pandas DataFrame
housing_df = housing.frame

print('Dataset shape:', housing_df.shape)

# Display first five rows
housing_df.head()
```

## Part 4: Exploratory Analysis

```python
# View available columns
housing_df.columns
```

```python
# Generate summary statistics
housing_df.describe()
```

## Part 5: Prepare Features and Target

For simplicity we will use:

- AveRooms
- HouseAge

To predict:

- MedHouseVal


```python
# Select input features
X = housing_df[['AveRooms', 'HouseAge']]

# Select target variable
y = housing_df['MedHouseVal']

print(X.head())
```

## Part 6: Split the Dataset

```python
# Create training and testing datasets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print('Training samples:', len(X_train))
print('Testing samples:', len(X_test))
```

## Part 7: Train a Machine Learning Model

```python
# Create a Linear Regression model
model = LinearRegression()

# Train the model
model.fit(X_train, y_train)

print('Training completed successfully.')
```

## Part 8: Evaluate the Model

```python
# Predict values for unseen data
predictions = model.predict(X_test)

# Calculate evaluation metrics
from sklearn.metrics import root_mean_squared_error

rmse = root_mean_squared_error(
    y_test,
    predictions
)
r2 = r2_score(y_test, predictions)

print(f'RMSE: {rmse:.4f}')
print(f'R2 Score: {r2:.4f}')
```

## Part 9: Save the Trained Model

```python
# Save the model to disk so that it can be reused
joblib.dump(model, 'model.joblib')

print('Model saved successfully.')
```

## Part 10: Load the Saved Model

```python
# Load a previously trained model
loaded_model = joblib.load('model.joblib')

print('Model loaded successfully.')
```

## Part 11: Build a Flask API

```python
# Import Flask components
from flask import Flask, request, jsonify

# Create Flask application
app = Flask(__name__)

# Load model during application startup
model = joblib.load('model.joblib')
```

### Health Endpoint

```python
# Health endpoint used to verify the API is running
@app.route('/health', methods=['GET'])
def health():

    return jsonify({
        'status': 'healthy',
        'model_loaded': True
    })
```

### Prediction Endpoint

```python
# Prediction endpoint
@app.route('/predict', methods=['POST'])
def predict():

    try:
        # Read JSON body
        data = request.get_json()

        if data is None:
            return jsonify({'error':'Request body is missing'}), 400

        # Required inputs
        required_fields = ['rooms', 'age']

        for field in required_fields:
            if field not in data:
                return jsonify({'error':f'Missing field: {field}'}), 400

        # Convert inputs to numbers
        rooms = float(data['rooms'])
        age = float(data['age'])

        # Validate inputs
        if rooms <= 0:
            return jsonify({'error':'Rooms must be greater than zero'}), 400

        if age < 0:
            return jsonify({'error':'Age cannot be negative'}), 400

        # Make prediction
        prediction = model.predict([[rooms, age]])

        return jsonify({
            'predicted_value': round(float(prediction[0]), 3)
        })

    except ValueError:
        return jsonify({'error':'Inputs must be numeric'}), 400

    except Exception as e:
        return jsonify({'error':str(e)}), 500
```

## Part 12: Run the API in Jupyter

```python
import threading
import time

# Function that starts the Flask server
def run_app():
    app.run(
        host='127.0.0.1',
        port=5000,
        debug=False,
        use_reloader=False
    )

# Launch Flask in a background thread
threading.Thread(target=run_app, daemon=True).start()

# Wait briefly for startup
time.sleep(2)

print('Flask API started successfully.')
```

## Part 13: Test the API
